# Example 06: Duration of strong shaking

Abrahamson and Silva (1996) give the duration of the ground motion as a function of the normalized Arias intensity, $I_a$: the time over which a fraction of the Arias intensity is accumulated. The model is evaluated for a scenario with a magnitude, a distance and a site condition. `interp` returns the duration for a normalized Arias intensity and a number of standard deviations from the median, which is used here to show the median and the $\pm 2$ standard-deviation range.

```python
import matplotlib.pyplot as plt
import numpy as np

import pygmm

%matplotlib inline
```

```python
scenario = pygmm.Scenario(mag=6, dist_rup=20, site_cond="soil")
```

```python
as96 = pygmm.AbrahamsonSilva1996(scenario)
as96.interp(0.75)
```

```python
nai = np.linspace(0, 1)
stds = [-2, 0, 2]
```

```python
durs = as96.interp(nai, stds)
```

```python
fig, ax = plt.subplots()

for std, _durs in zip(stds, durs):
    ax.plot(_durs, nai, label=f"{std:.1f}")

ax.legend()
ax.set(xlabel="Duration (sec)", ylabel="Normalized $I_a$")
fig.tight_layout()
fig;
```
